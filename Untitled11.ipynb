```python
# CELDA 0.1 · Importaciones y parámetros globales
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans, DBSCAN
from sklearn.metrics import silhouette_score
from sklearn.neighbors import NearestNeighbors
from sklearn.decomposition import PCA

warnings.filterwarnings("ignore")

# --- Parámetros (un solo lugar para cambiar cualquier decisión) ---
SEMILLA      = 42
RUTA_DATOS   = "dataset_clientes_financieros.xlsx"
HOJA         = "Datos_Transaccionales"
COL_ID       = "ID_Cliente"
COL_INGRESO  = "Ingreso_Anual_USD"
COL_MORA     = "Dias_Mora_Historico"

np.random.seed(SEMILLA)
sns.set_theme(style="whitegrid", palette="deep")
plt.rcParams["figure.figsize"] = (10, 5)
```

```python
# CELDA 1.1 · Carga e inspección inicial
def cargar_datos(ruta: str, hoja: str) -> pd.DataFrame:
    """Lee la hoja de Excel y devuelve un DataFrame."""
    return pd.read_excel(ruta, sheet_name=hoja)


def inspeccion_inicial(df: pd.DataFrame) -> None:
    """Imprime forma, tipos, nulos, duplicados y primeras filas."""
    print(f"Filas: {df.shape[0]} | Columnas: {df.shape[1]}\n")
    print("--- Tipos de datos ---")
    print(df.dtypes, "\n")
    print("--- Valores nulos por columna ---")
    print(df.isna().sum(), "\n")
    print(f"Filas duplicadas: {df.duplicated().sum()}")
    print(f"IDs únicos: {df[COL_ID].is_unique}\n")
    display(df.head())


df = cargar_datos(RUTA_DATOS, HOJA)
COLUMNAS_NUM = [c for c in df.columns if c != COL_ID]   # variables analíticas
inspeccion_inicial(df)
```

```python
# CELDA 1.2 · Estadísticas descriptivas
display(df[COLUMNAS_NUM].describe().T.round(2))
print("\nAsimetría (skew) por variable:")
print(df[COLUMNAS_NUM].skew().round(2).sort_values(ascending=False))
```

```python
# CELDA 1.3 · Histogramas de todas las variables
def graficar_distribuciones(df, columnas, ncols=3):
    nrows = int(np.ceil(len(columnas) / ncols))
    fig, axes = plt.subplots(nrows, ncols, figsize=(5 * ncols, 4 * nrows))
    axes = axes.ravel()
    for ax, col in zip(axes, columnas):
        sns.histplot(df[col], kde=True, bins=30, ax=ax)
        ax.axvline(df[col].mean(), color="red", ls="--", label="Media")
        ax.axvline(df[col].median(), color="green", ls="--", label="Mediana")
        ax.set_title(f"{col}\nAsimetría = {df[col].skew():.2f}")
        ax.legend()
    for ax in axes[len(columnas):]:
        ax.axis("off")
    plt.tight_layout()
    plt.show()


graficar_distribuciones(df, COLUMNAS_NUM)
```

```python
# CELDA 1.4 · Boxplots para ver outliers
def graficar_boxplots(df, columnas, ncols=3):
    nrows = int(np.ceil(len(columnas) / ncols))
    fig, axes = plt.subplots(nrows, ncols, figsize=(5 * ncols, 3.5 * nrows))
    axes = axes.ravel()
    for ax, col in zip(axes, columnas):
        sns.boxplot(x=df[col], ax=ax, color="lightsteelblue")
        ax.set_title(col)
    for ax in axes[len(columnas):]:
        ax.axis("off")
    plt.tight_layout()
    plt.show()


graficar_boxplots(df, COLUMNAS_NUM)
```

```python
# CELDA 1.5 · Mapa de calor de correlación
def graficar_heatmap(df, columnas):
    corr = df[columnas].corr()
    plt.figure(figsize=(9, 7))
    sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm",
                vmin=-1, vmax=1, square=True, linewidths=0.5)
    plt.title("Matriz de correlación (Pearson)")
    plt.show()
    return corr


corr = graficar_heatmap(df, COLUMNAS_NUM)

# Pares fuertemente correlacionados (|r| >= 0.7), sin duplicar
pares = (corr.where(np.triu(np.ones(corr.shape, dtype=bool), k=1))
             .stack().rename("r").reset_index())
pares.columns = ["Variable_A", "Variable_B", "r"]
print("Pares con |r| >= 0.70:")
display(pares[pares["r"].abs() >= 0.70].sort_values("r", ascending=False))
```

```python
# CELDA 1.6 · Comparación de perfiles según nivel de mora
UMBRAL_MORA = 30   # convención de la industria: >30 días = mora relevante
grupo_mora = np.where(df[COL_MORA] > UMBRAL_MORA, "Mora > 30 días", "Mora <= 30 días")

print("Tamaño de cada grupo:")
print(pd.Series(grupo_mora).value_counts(), "\n")

comparacion = df.groupby(grupo_mora)[COLUMNAS_NUM].mean().round(2).T
display(comparacion)

print("\nCorrelación de cada variable con los días de mora:")
print(df[COLUMNAS_NUM].corr()[COL_MORA].drop(COL_MORA).sort_values().round(2))
```

```python
# CELDA 2.1 · Selección de features y escalado
def preparar_datos(df, columnas):
    """Devuelve (X_original, X_escalado, scaler)."""
    X = df[columnas].copy()
    scaler = StandardScaler()
    X_esc = pd.DataFrame(scaler.fit_transform(X), columns=columnas, index=X.index)
    return X, X_esc, scaler


FEATURES = COLUMNAS_NUM          # ID_Cliente ya quedó fuera
X, X_scaled, scaler = preparar_datos(df, FEATURES)

print("Verificación del escalado (media ≈ 0, desviación ≈ 1):")
display(X_scaled.describe().loc[["mean", "std"]].round(3))
```

```python
# CELDA 3.1 · Método del Codo + Coeficiente de Silueta
def evaluar_k(X, k_min=2, k_max=10):
    filas = []
    for k in range(k_min, k_max + 1):
        km = KMeans(n_clusters=k, n_init=10, random_state=SEMILLA).fit(X)
        filas.append({"K": k,
                      "Inercia": km.inertia_,
                      "Silueta": silhouette_score(X, km.labels_)})
    return pd.DataFrame(filas)


def graficar_evaluacion_k(res):
    fig, ax = plt.subplots(1, 2, figsize=(14, 5))
    ax[0].plot(res["K"], res["Inercia"], "o-")
    ax[0].set_title("Método del Codo (Inercia)")
    ax[0].set_xlabel("K"); ax[0].set_ylabel("Inercia (WCSS)")
    ax[1].plot(res["K"], res["Silueta"], "o-", color="darkorange")
    ax[1].set_title("Coeficiente de Silueta")
    ax[1].set_xlabel("K"); ax[1].set_ylabel("Silueta promedio")
    plt.tight_layout()
    plt.show()


resultados_k = evaluar_k(X_scaled)
display(resultados_k.round(3))
graficar_evaluacion_k(resultados_k)

K_OPTIMO = int(resultados_k.loc[resultados_k["Silueta"].idxmax(), "K"])
print(f"K con mayor silueta: {K_OPTIMO}")
```

```python
# CELDA 3.2 · Entrenamiento final de K-Means y perfilado
kmeans = KMeans(n_clusters=K_OPTIMO, n_init=10, random_state=SEMILLA)
etiquetas_km = kmeans.fit_predict(X_scaled)

df_final = df.copy()
df_final["Cluster_KMeans"] = etiquetas_km


def perfilar_clusters(df, col_cluster, columnas):
    """Promedio de cada variable por cluster + tamaño."""
    perfil = df.groupby(col_cluster)[columnas].mean().round(2)
    perfil.insert(0, "N_Clientes", df[col_cluster].value_counts().sort_index())
    perfil.insert(1, "Pct_Cartera", (perfil["N_Clientes"] / len(df) * 100).round(1))
    return perfil


perfil_km = perfilar_clusters(df_final, "Cluster_KMeans", FEATURES)
display(perfil_km)
```

```python
# CELDA 3.3 · Visualización de los clusters
# (a) Perfil relativo: cuánto se desvía cada cluster del promedio global (en desviaciones estándar)
perfil_z = (perfil_km[FEATURES] - df[FEATURES].mean()) / df[FEATURES].std()
plt.figure(figsize=(10, 4))
sns.heatmap(perfil_z, annot=True, fmt=".2f", cmap="RdYlGn_r", center=0)
plt.title("Perfil relativo de cada cluster (z-score vs. promedio global)")
plt.show()

# (b) Proyección a 2D con PCA
pca = PCA(n_components=2, random_state=SEMILLA)
coords = pca.fit_transform(X_scaled)
plt.figure(figsize=(9, 6))
sns.scatterplot(x=coords[:, 0], y=coords[:, 1], hue=etiquetas_km,
                palette="Set1", s=40, alpha=0.8)
plt.title(f"K-Means (K={K_OPTIMO}) proyectado con PCA "
          f"(varianza explicada: {pca.explained_variance_ratio_.sum():.1%})")
plt.xlabel("PC1"); plt.ylabel("PC2")
plt.show()
```

```python
# CELDA 3.4 · Calibración de eps con el gráfico de k-distancias
MIN_SAMPLES = 2 * len(FEATURES)     # regla práctica: 2 x número de dimensiones


def grafico_k_distancias(X, k):
    vecinos = NearestNeighbors(n_neighbors=k).fit(X)
    distancias, _ = vecinos.kneighbors(X)
    d_k = np.sort(distancias[:, -1])            # distancia al k-ésimo vecino
    plt.figure(figsize=(10, 5))
    plt.plot(d_k)
    plt.title(f"Gráfico de k-distancias (k = {k})")
    plt.xlabel("Clientes ordenados por distancia")
    plt.ylabel(f"Distancia al vecino #{k}")
    plt.grid(True)
    plt.show()
    return d_k


d_k = grafico_k_distancias(X_scaled, MIN_SAMPLES)
print("Percentiles de la distancia:")
print(pd.Series(d_k).quantile([.5, .8, .9, .95, .99]).round(2))
```

```python
# CELDA 3.5 · Experimentación con eps y min_samples
def explorar_dbscan(X, lista_eps, lista_ms):
    filas = []
    for ms in lista_ms:
        for eps in lista_eps:
            etiquetas = DBSCAN(eps=eps, min_samples=ms).fit_predict(X)
            filas.append({"min_samples": ms, "eps": eps,
                          "N_Clusters": len(set(etiquetas)) - (1 if -1 in etiquetas else 0),
                          "N_Ruido": int((etiquetas == -1).sum())})
    return pd.DataFrame(filas)


grid = explorar_dbscan(X_scaled,
                       lista_eps=[0.6, 0.8, 1.0, 1.2, 1.5, 1.8, 2.0],
                       lista_ms=[5, 8, MIN_SAMPLES])
display(grid.pivot(index="eps", columns="min_samples", values="N_Ruido")
            .rename_axis(columns="N_Ruido según min_samples"))
```

```python
# CELDA 3.6 · Modelo DBSCAN final
EPS = 1.8      # decisión basada en CELDAS 3.4 y 3.5

dbscan = DBSCAN(eps=EPS, min_samples=MIN_SAMPLES)
etiquetas_db = dbscan.fit_predict(X_scaled)
df_final["Cluster_DBSCAN"] = etiquetas_db

n_clusters = len(set(etiquetas_db)) - (1 if -1 in etiquetas_db else 0)
n_ruido = int((etiquetas_db == -1).sum())
print(f"Clusters encontrados : {n_clusters}")
print(f"Clientes ruido (-1)  : {n_ruido} ({n_ruido / len(df_final):.1%} de la cartera)")
```

```python
# CELDA 3.7 · Análisis de los clientes catalogados como ruido
ruido  = df_final[df_final["Cluster_DBSCAN"] == -1]
normal = df_final[df_final["Cluster_DBSCAN"] != -1]

comparacion_db = pd.DataFrame({
    "Media_Ruido":  ruido[FEATURES].mean(),
    "Media_Normal": normal[FEATURES].mean(),
})
comparacion_db["Veces_vs_Normal"] = (comparacion_db["Media_Ruido"] /
                                      comparacion_db["Media_Normal"]).round(2)
display(comparacion_db.round(2))

print("\nDetalle de cada cliente anómalo:")
display(ruido)

print("\nz-scores (desviaciones estándar respecto al promedio global):")
display(X_scaled.loc[ruido.index].round(2))
```

```python
# CELDA 3.8 · Visualización DBSCAN y cruce con K-Means
plt.figure(figsize=(9, 6))
es_ruido = etiquetas_db == -1
plt.scatter(coords[~es_ruido, 0], coords[~es_ruido, 1], c="lightgray", s=30, label="Clientes normales")
plt.scatter(coords[es_ruido, 0], coords[es_ruido, 1], c="red", s=90, marker="X", label="Ruido (anomalías)")
plt.title("DBSCAN: anomalías detectadas (proyección PCA)")
plt.xlabel("PC1"); plt.ylabel("PC2"); plt.legend()
plt.show()

print("Cruce K-Means vs DBSCAN (¿en qué cluster comercial cayeron los anómalos?):")
display(pd.crosstab(df_final["Cluster_KMeans"], df_final["Cluster_DBSCAN"]))
```

```python
# CELDA 4.1 · Exportar resultados
df_final.to_csv("clientes_segmentados.csv", index=False)
ruido.to_csv("clientes_anomalos_riesgo.csv", index=False)
print("Archivos generados: clientes_segmentados.csv, clientes_anomalos_riesgo.csv")
```
